```python
def sendArray(comport, baudrate, array, csvPath: str = None, readTimeout: int = 5, sleepDur: float = 0.001):
    """
    Send a single list of values over a serial connection to an Arduino.

    Parameters
    ----------
    comport : str
        The serial port to use.
    baudrate : int
        The baudrate to use.
    array : array-like
        The array of values to send.
    csvPath : str, optional
        The path to write the CSV file to. The default is None.

    Returns
    -------
    None
    """
    ser = serial.Serial(comport, baudrate, timeout=0.1)
    for item in array:
        message = str(item) + ' '
        writeSerial(ser, message)
        print(f'{message} gesendet!')
        time.sleep(sleepDur)
    readSerial(ser, csvPath, readTimeout)

def writeSerial(ser, message: str):
    ser.write(bytes(message, 'utf-8'))

def readSerial(ser, csvPath: str = None, timeout: int = 5):
    """
    Read data from a serial connection and print or write it to a CSV file.

    Parameters
    ----------
    ser : serial.Serial
        The serial connection to read from.
    csvPath : str, optional
        The path to write the CSV file to. The default is None.
    timeout : int, optional
        The timeout in seconds for the serial connection. The default is 5.

    Returns
    -------
    None
    """
    start = time.time()
    while True:
        data = ser.readline().decode()
        if data:
            if csvPath != None:
                with open(csvPath + 'inoCapture.csv', 'a') as f:
                    f.write(data + '\n')
            else:
                print(data)
            break
        end = time.time()
        if end - start > timeout:
            print("Timeout!")
            break
```

Nimmt eine CSV und gibt jeweils die größte, die kleinste und die Zahl mit den meisten Nachkommastellen zurück

```python
import csv

def count_decimal_places(number: str) -> int:
    """Gibt die Anzahl der Nachkommastellen einer Zahl im String-Format zurück."""
    if '.' in number:
        return len(number.split('.')[1])
    return 0

def analyze_csv(file_path: str):
    smallest = float('inf')
    largest = float('-inf')
    most_decimals = ""
    max_decimal_count = -1

    with open(file_path, mode='r') as csv_file:
        csv_reader = csv.reader(csv_file)
        
        # Überspringe die erste Zeile (Header)
        next(csv_reader)
        
        for row in csv_reader:
            for num_str in row:
                try:
                    num = float(num_str)  # Konvertiere den String in eine Zahl
                    # Finde die kleinste Zahl
                    if num < smallest:
                        smallest = num
                    # Finde die größte Zahl
                    if num > largest:
                        largest = num
                    # Zähle die Nachkommastellen
                    decimal_count = count_decimal_places(num_str)
                    if decimal_count > max_decimal_count:
                        max_decimal_count = decimal_count
                        most_decimals = num_str
                except ValueError:
                    # Falls die Konvertierung in float nicht klappt, ignoriere den Wert
                    continue
    
    return smallest, largest, most_decimals, max_decimal_count

# Pfad zur CSV-Datei anpassen
file_path = 'datasets/harus/xtest.csv'
smallest, largest, most_decimals, max_decimal_count = analyze_csv(file_path)
print("Die kleinste Zahl:", smallest)
print("Die größte Zahl:", largest)
print("Die Zahl mit den meisten Nachkommastellen:", most_decimals)
print(max_decimal_count)
decimal = float(most_decimals)
decimal_dec = f"{decimal:.20f}"
erste = int(decimal*(10**7))
zweite = int((decimal*(10**7) - erste)*(10**7))
print(f'Zahl mit den meisten Nachkommastellen: {decimal}')
print(f'Zahl mit den meisten Nachkommastellen: {decimal_dec}')

print(erste)
print(zweite)
zusammengesetzt = (erste/(10**7))+(zweite/(10**14))
zusammengesetzt_dec = f"{zusammengesetzt:.14f}"
print(decimal_dec)
print(zusammengesetzt_dec)

```

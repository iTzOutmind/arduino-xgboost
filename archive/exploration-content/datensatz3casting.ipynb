```python
from ucimlrepo import fetch_ucirepo 
import pandas as pd

import pandas as pd
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
# Diabetes Datensatz

# fetch dataset 
cdc_diabetes_health_indicators = fetch_ucirepo(id=891) 
  
# data (as pandas dataframes) 
X_diab = cdc_diabetes_health_indicators.data.features 
y_diab = cdc_diabetes_health_indicators.data.targets 
```

```python
# Künstlicher Datensatz Mushrooms

# fetch dataset 
secondary_mushroom = fetch_ucirepo(id=848) 
  
# data (as pandas dataframes) 
X_mushy = secondary_mushroom.data.features 
y_mushy = secondary_mushroom.data.targets 
```

```python
# Android Malware

# fetch dataset 
naticusdroid_android_permissions = fetch_ucirepo(id=722) 
  
# data (as pandas dataframes) 
X_android = naticusdroid_android_permissions.data.features 
y_android = naticusdroid_android_permissions.data.targets 
  
```

```python
# presence

data1 = pd.read_csv("datasets/presence/pirvision_office_dataset1.csv")
data2 = pd.read_csv("datasets/presence/pirvision_office_dataset2.csv")
```

```python
# Mushy Modell

x_mushy_cat = list(X_mushy.columns)
[x_mushy_cat.remove(i) for i in ["cap-diameter","stem-height","stem-width"]]

encoders = {} # Dict that stores columns and corresponding Label Encoders

for i in x_mushy_cat:
    le = LabelEncoder()
    le.fit(X_mushy[i])
    X_mushy[i] = le.transform(X_mushy[i])
    encoders.update({i:le})

le_y_mushy = LabelEncoder()
y_mushy['class'] = le.fit_transform(y_mushy)


xtrainm, xtestm, ytrainm, ytestm = train_test_split( # Split into training and test datasets
    X_mushy,
    y_mushy,
    test_size=0.33,
    random_state=0,
    stratify=y_mushy
)

mushy_clf = XGBClassifier()


evalset_mushy = [(xtrainm,ytrainm),(xtestm,ytestm)]

mushy_clf.set_params(
    objective = "binary:logistic",
    tree_method = 'exact',
)

mushy_clf.fit(
    xtrainm, ytrainm,
    eval_set = evalset_mushy,
    verbose = 0
)

yhat_mushy = mushy_clf.predict(xtestm)

accsc_mushy = accuracy_score(ytestm, yhat_mushy)

```

```python
# Diabetes Modell

xtraind, xtestd, ytraind, ytestd = train_test_split( # Split into training and test datasets
    X_diab,
    y_diab,
    test_size=0.33,
    random_state=0,
    stratify=y_diab
)

diab_clf = XGBClassifier()

evalset_diab = [(xtraind,ytraind),(xtestd,ytestd)]

diab_clf.set_params(
    objective = "binary:logistic",
    tree_method = 'exact'
)

diab_clf.fit(
    xtraind, ytraind,
    eval_set = evalset_diab,
    verbose = 0
)

yhatd = diab_clf.predict(xtestd)

acc_diab = accuracy_score(ytestd, yhatd)
```

```python
# Android Malware Detection

xtraina, xtesta, ytraina, ytesta = train_test_split( # Split into training and test datasets
    X_android,
    y_android,
    test_size=0.33,
    random_state=0,
    stratify=y_android
)

android_clf = XGBClassifier()

evalset_android = [(xtraina,ytraina),(xtesta,ytesta)]

android_clf.set_params(
    objective = "binary:logistic",
    tree_method = 'exact'
)

android_clf.fit(
    xtraina, ytraina,
    eval_set = evalset_android,
    verbose = 0
)

yhat_android = android_clf.predict(xtesta)

acc_andr = accuracy_score(ytesta, yhat_android)
```

```python
# Presence Sensor Modell

xdata_presence = data2.iloc[:,3:]
ydata_presence = data2.iloc[:,2]

le_presence = LabelEncoder()
ydata_presence = le.fit_transform(ydata_presence)

xtrainp, xtestp, ytrainp, ytestp = train_test_split( # Split into training and test datasets
    xdata_presence,
    ydata_presence,
    test_size=0.33,
    random_state=0,
    stratify=ydata_presence
)

presence_clf = XGBClassifier()

evalset_presence = [(xtrainp,ytrainp),(xtestp,ytestp)]

presence_clf.set_params(
    objective = "binary:logistic",
    tree_method = 'exact'
)

presence_clf.fit(
    xtrainp, ytrainp,
    eval_set = evalset_presence,
    verbose = 0
)

yhat_presence = presence_clf.predict(xtestp)

acc_presence = accuracy_score(ytestp, yhat_presence)
```

```python
print(f'Diabetes: {acc_diab}')
print(f'Mushrooms: {accsc_mushy}')
print(f'Android Malware: {acc_andr}')
print(f'Presence Detection: {acc_presence}')
```

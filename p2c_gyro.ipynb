## 1 Module

```python
import numpy as np
import pandas as pd

from xgboost import XGBClassifier

import sklearn.metrics as skm
from sklearn.model_selection import train_test_split

import m2cgen as m2c
```

# 2 Modellimport

```python
# Parameters identical to clf_final in basemodell_gyro.ipynb

clf = XGBClassifier(objective = "binary:logistic",
                    tree_method = 'exact',
                    n_estimators = 854,
                    max_depth = 9)
clf.load_model("model-exports/base_gyro.ubj")
```

# 3 Set base_score

```python
clf.set_params(base_score = float(clf.base_score))

print(f'Set parameter "base_score" to: {float(clf.base_score)}')
```

# 4 Funktionen

```python
def portToC(model):
    path = 'model-exports/plainC/'
    filename = 'base_gyro.c'

    with open(path + filename,'w') as f:
        code = m2c.export_to_c(model)
        f.write(code)

    print(f'Model exported to: "{path + filename}"')
```

# 5 Porting to C

```python
portToC(clf)
```

- **boolean**: True or False

- **unsigned char**: Same as byte -> Shouldn't be used

- **word**: Unsigned Number from 0 to 65535
- **unsigned int**: Sane as word -> Shouldn't be used

- **unsigned long**: Unsigned number from 0 to 4.294.967.295
- **long**: Signed number from -2.147.483.648 to 2.147.483.647\

Genauigkeit: 7 Dezimalstellen (vor und nach dem Komma!)

## <a name="featureselection" style="text-decoration: none; color: white;">Feature Selection</a>

### Links

- [sklearn.feature_selection](https://scikit-learn.org/stable/api/sklearn.feature_selection.html)

### Methoden

- **Threshold 0**\
Es werden die Feature_Importances aller Features aus dem Modell ermittelt und alle Features mit einem Wert von 0.0 verworfen.
- **Iterativ**\
Es werden wieder die Feature_Importances aller Features aus dem Modell ermittelt. Es wird ein Grenzwert festgelegt, nach dem entschieden wird, ob ein Feature noch im Datensatz verbleiben soll oder nicht. Dieser Grenzwert wird schrittweise so lange erhöht, bis ein Abbruchkriterium erreicht wurde. Als Abbruchkriterien wurden hier eine Zielmenge an verbleibenden Features oder ein maximaler erlaubter Verlust an Modell-Accuracy verwendet. Hierzu gab es 2 Vorgehensweisen\
    1. Zuerst wird 0.0 als Grenzwert gesetzt und alle Features mit einer Importance von bis zu diesem Grenzwert werden aus dem Datensatz entfernt. Der Grenzwert wird durch einen konstaten Wert erhöht und der Vorgang wird wiederholt.
    2. Aus den zuvor ermittelten Importances wird eine Liste angefertigt und alle redundanten Werte/Duplikate werden aus dieser Liste entfernt. Die Liste wird zunächst nach aufsteigender Reihenfolge sortiert. Im Anschluss wird über diese Liste iteriert, wobei der jeweilige Listeneintrag als aktueller Grenzwert gilt. Alle Features mit einer Importance bis zu diesem Grenzwert werden entfernt.
- **Recursive Feature Elimintation** (RFE)\
Rekursives Ermitteln zu entfernender Features, wobei zunächst ein Modell mit allen Features trainiert wird und rekursiv kleiner werdende Sets an Features erzeugt werden. Ziel ist die Ermittlung der bestmöglichen Kombination aus verbleibenden Features -> Hierzu war bereits eine Funktion im Scikit-Learn-Paket enthalten.
- [**RFECV**](https://scikit-learn.org/stable/modules/generated/sklearn.feature_selection.RFECV.html) -> TODO\
Eine erweiterte Form vom RFE, bei der mithilfe von Cross-Validation auch die optimale Menge an Features ermittelt wird

### Methoden

#### 1. Grid Search

Es werden für eine Reihe an Hyperparametern Werte festgelegt, die für die Modellerzeugung angenommen werden können.
Für jede mögliche Kombination dieser Werte/Hyperparameter wird jeweils ein Modell trainiert, um die optimale Kombination zu ermitteln.
Diese Methode ist einfach zu implementieren, eignet sich jedoch auch nur für eine kleine Anzahl an Parametern bzw. Werten, da 
die Suche schnell sehr rechenintensiv werden kann.

#### 2. Random(ized) Search

Random Search ist eine Abwandlung/Variation von Grid Search, wobei nicht mehr gezielt Werte, sondern Intervalle für Hyperparameter festgelegt werden.
Werte innerhalb dieser Intervalle werden nach Zufall gewählt und damit Modelle trainiert, statt jede mögliche Kombination zu parametrisieren.
Eine Suche mit Random Search kann effizienter sein als Grid Search, jedoch wird auch diese Methode mit steigender Anzahl an Parametern und wachsenden
Intervallen schnell sehr rechenintensiv.\
Parameterwerte müssen nicht zwangsläufig aus Intervallen gezogen werden. Auch hier können diskrete Werte im Vorfeld festgelegt werden, aus denen gezogen werden kann. Im Gegensatz zu Grid Search werden hier jedoch nicht zwangsläufig alle Kombinationen durchprobiert.

#### 3. Bayesian Optimization

Bei der Bayes'schen Optimierung handelt es sich um eine ausgeklügeltere Methode, welche Bayes'sche Methoden nutzt um eine Funktion zu modellieren, welche die untersuchten Hyperparameter in Relation zu der Modellperformance setzt. Diese Optimierungsmethode ist effizienter als Grid oder Random Search, da sie versucht Exploration und Exploitation des Suchraums im Gleichgewicht zu halten. BO kann dabei mit einer größeren Anzahl an Parametern arbeiten. Im Gegenzug ist eine korrekte Implementierung von BO aufwändiger und kann mehr Rechenleistung in Anspruch nehmen.

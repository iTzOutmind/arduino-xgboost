# 2 Funktion definieren

```python
def portToC(model, filename):
    path = 'model-exports/plainC/'

    with open(path + filename,'w') as f:
        code = m2c.export_to_c(model)
        f.write(code)

    print(f'Model exported to: "{path + filename}"')
```

# 4 Feature Selection Embedded portieren

```python
clf_fs_emb = XGBClassifier(
    objective = "binary:logistic",
    tree_method = 'exact',
    n_estimators = 541,
    max_depth = 1
)
clf_fs_emb.load_model("model-exports/aktuelle-exports/fs_embedded_agmp.json")

clf_fs_emb.set_params(base_score = float(clf_fs_emb.base_score))
print(f'Set parameter "base_score" to: {float(clf_fs_emb.base_score)}')

portToC(clf_fs_emb, 'fs_embedded_agmp.c')
```

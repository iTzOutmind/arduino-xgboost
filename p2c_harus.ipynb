# 1 Module

```python
from xgboost import XGBClassifier

import m2cgen as m2c
```

# 2 Funktionen definieren

```python
def portToC(model, filename):
    path = 'model-exports/plainC/'

    with open(path + filename,'w') as f:
        code = m2c.export_to_c(model)
        f.write(code)

    print(f'Model exported to: "{path + filename}"')
```

# 4 Feature Selection Embedded portieren

```python
clf_fs_emb = XGBClassifier(
    objective = "multi:softmax",
    tree_method = 'exact',
    n_estimators = 596,
    max_depth = 6
)
clf_fs_emb.load_model("model-exports/aktuelle-exports/fs_embedded_harus.json")

clf_fs_emb.set_params(
    base_score = float(clf_fs_emb.base_score),
    num_parallel_tree = 1
)
print(f'Set parameter "base_score" to: {float(clf_fs_emb.base_score)}')

portToC(clf_fs_emb, 'fs_embedded_harus.c')
```

# 5 Feature Selection RFECV portieren

```python
clf_fs_rfecv = XGBClassifier(
    objective = "multi:softmax",
    tree_method = 'exact',
    n_estimators = 336,
    max_depth = 10
)
clf_fs_rfecv.load_model("model-exports/aktuelle-exports/fs_rfecv_harus.json")

clf_fs_rfecv.set_params(
    base_score = float(clf_fs_emb.base_score),
    num_parallel_tree = 1
)
print(f'Set parameter "base_score" to: {float(clf_fs_rfecv.base_score)}')

portToC(clf_fs_rfecv, 'fs_rfecv_harus.c')
```

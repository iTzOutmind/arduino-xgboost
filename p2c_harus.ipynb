# 1 Module

```python
from xgboost import XGBClassifier

import m2cgen as m2c
```

# 2 Modellimport

```python
# Parameters identical to clf_final in basemodell_gyro.ipynb

clf = XGBClassifier(objective = "binary:logistic",
                    tree_method = 'exact',
                    n_estimators = 194,
                    max_depth = 4)
clf.load_model("model-exports/aktuelle-exports/base_harus_neu.json")
```

# 3 Set required params for m2cgen

```python
clf.set_params(base_score = float(clf.base_score),
               num_parallel_tree = 1 # identical to json-export
               )

print(f'Set parameter "base_score" to: {float(clf.base_score)}')
```

# 4 Funktionen

```python
def portToC(model):
    path = 'model-exports/plainC/'
    filename = 'base_harus' \
    '.c'

    with open(path + filename,'w') as f:
        code = m2c.export_to_c(model)
        f.write(code)

    print(f'Model exported to: "{path + filename}"')
```

# 5 Porting to C

```python
portToC(clf)
```

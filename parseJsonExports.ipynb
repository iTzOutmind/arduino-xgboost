```python
import json
```

```python
def getNodesTotal(filename: str, directory: str = 'model-exports/aktuelle-exports/'):
    path = directory + filename
    nodesTotal = 0

    with open(path, 'r', encoding='utf-8') as datei:
        jsonData = json.load(datei)
    
    for tree in jsonData['learner']['gradient_booster']['model']['trees']:
        nodesTotal += int(tree['tree_param']['num_nodes'])

    return(nodesTotal)
```

```python
nodesAgmp = getNodesTotal("base_agmp.json")
nodesHarus = getNodesTotal("base_harus.json")
nodesSeMu = getNodesTotal("base_semu.json")
print("##### Basismodelle #####\n")
print(f'Total nodes in AGMP: \t {nodesAgmp}\t| max_depth: 1\t| n_estimators: 1314')
print(f'Total nodes in Harus: \t {nodesHarus}\t| max_depth: 4\t| n_estimators: 1164 (194 * 6)')
print(f'Total nodes in Mushroom: {nodesSeMu}\t| max_depth: 7\t| n_estimators: 551 \n')

print(f'Theoretical size in bytes (Double): {nodesAgmp * 8} Bytes')
print(f'Theoretical size in bytes (Double): {nodesHarus * 8} Bytes')
print(f'Theoretical size in bytes (Double): {nodesSeMu * 8} Bytes')
```

```python
import json
```

```python
def getNodesTotal(filename: str, directory: str = 'model-exports/aktuelle-exports/'):
    path = directory + filename
    nodesTotal = 0

    with open(path, 'r', encoding='utf-8') as datei:
        jsonData = json.load(datei)
    
    for tree in jsonData['learner']['gradient_booster']['model']['trees']:
        nodesTotal += int(tree['tree_param']['num_nodes'])

    return(nodesTotal)
```

```python
nodesGyro = getNodesTotal("base_gyro_neu.json")
nodesHarus = getNodesTotal("base_harus_neu.json")
nodesMushroom = getNodesTotal("base_mushroom_neu.json")

print(f'Total nodes in AGMP: \t {nodesGyro}\t| max_depth: 1\t| n_estimators: 1314')
print(f'Total nodes in Harus: \t {nodesHarus}\t| max_depth: 4\t| n_estimators: 1164 (194 * 6)')
print(f'Total nodes in Mushroom: {nodesMushroom}\t| max_depth: 7\t| n_estimators: 551 \n')

print(f'Theoretical size in bytes (Double): {nodesGyro * 8} Bytes')
print(f'Theoretical size in bytes (Double): {nodesHarus * 8} Bytes')
print(f'Theoretical size in bytes (Double): {nodesMushroom * 8} Bytes')

```
